```python
# import numpy as np
a1 = np.random.random((3,3))
a1 = np.round(a1*100)
a1
```

```python
# max/min/sum/prod
# 0 > col and 1 > row
# np.max(a1,axis=1)
# np.max(a1,axis=0)
# np.sum(a1,axis=1)
np.prod(a1,axis=0)
```

```python
# mean/median/std/var
np.mean(a1,axis=1)
```

```python
# trigonomoetric functions
np.sin(a1)
```

```python
# dot product
a2 = np.arange(12).reshape(3,4)
a3 = np.arange(12,24).reshape(4,3)

np.dot(a2,a3)
```
